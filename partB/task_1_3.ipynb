# Task 1.3 — What the Paper Claims to Improve (9 marks)

**Paper**: *Learning to Grade Short Answer Questions using Semantic Similarity Measures and Dependency Graph Alignments* (Mohler, Bunescu & Mihalcea, ACL 2011)

---

## Main Baseline (2 marks)

The paper compares against **individual bag-of-words (BOW) semantic similarity measures used in isolation**, specifically the approach from **Mohler & Mihalcea (2009)**, which assigns a grade based on a single measure of text-to-text semantic similarity between the student and instructor answers. The strongest individual baselines include the **Lesk** measure (Pearson ρ = 0.462 with question demoting) and **JCN** (ρ = 0.461), as shown in Table 5. Additionally, the paper treats the **average grade baseline** (assigning the mean training grade to all answers, yielding RMSE = 1.097) and the standalone **tf*idf cosine similarity** (ρ = 0.327) as reference points.

## Limitation of the Baseline (2 marks)

The key limitation the paper identifies is that **no single similarity measure captures all relevant aspects of answer quality**. Each individual measure has blind spots: WordNet-based measures miss distributional similarity, corpus-based measures (LSA, ESA) miss fine-grained semantic relations, and tf*idf misses synonymy entirely. Using any one measure in isolation means discarding complementary information from other measures. The paper also argues that pure BOW approaches fundamentally ignore **word order and syntactic structure** — they cannot distinguish between "dog bites man" and "man bites dog" (Section 2, discussing C-Rater's motivation).

## How the Proposed Method Overcomes This (1 mark)

The paper overcomes this limitation by **combining multiple heterogeneous feature types using supervised learning**. Specifically, it concatenates 8 graph alignment features ($\psi_G$) with 11+ BOW semantic similarity features ($\psi_B$) into a joint 30-dimensional feature vector, and trains an SVR (or SVMRank) model with a linear kernel to learn the optimal weighting of these features from labelled training data (Section 3.4). This allows the model to leverage complementary strengths of different measures — for instance, using structural alignment where word order matters and BOW similarity where synonymy is important.

## When the Paper's Method Would NOT Outperform the Baseline (4 marks)

The paper's method would likely **not** outperform the simple BOW baseline in scenarios where **student answers are heavily paraphrased with very low lexical overlap but are semantically correct**. Since even the combined model relies substantially on surface-level and distributional similarity features (the BOW component $\psi_B$ contributes 11 of the 30 features, and both tf*idf and word overlap are core surface measures), deeply paraphrased answers would receive low similarity scores across most features. The dependency graph alignment might partially help, but the paper's own node-matching perceptron uses WordNet and LSA-based features (Table 2, Semantic row: 36 features) which still assume some vocabulary overlap. Additionally, on a **highly balanced dataset** (equal proportion of each grade level), the paper acknowledges that the tf*idf baseline's tendency to cluster predictions around the mean grade (Section 6, "46% of the scores output by the tf*idf measure were within the 4 to 4.5 range") would become a disadvantage, and the relative gain from combining features via SVR might be smaller because the skew-exploitation effect disappears.
